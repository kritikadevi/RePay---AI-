```python
# Drop columns that should not be used as features
X = df.drop(
    columns=[
        "payment_id",
        "recovered",
        "recovered_amount"
    ]
)

# Target
y = df["recovered"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
#  One-Hot Encoding.

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Categorical columns
categorical_cols = [
    "payment_method",
    "failure_reason",
    "recovery_action"
]

# Numerical columns
numerical_cols = [
    "amount",
    "previous_successful_payments",
    "previous_failed_payments",
    "retry_count",
    "customer_success_rate",
    "hour_of_payment",
    "customer_tenure_days"
]

# Encoder
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        ),
        (
            "num",
            "passthrough",
            numerical_cols
        )
    ]
)
```

```python
def choose_best_action(payment_details):
    
    # Actions our AI agent can choose
    actions = ["Retry", "UPI Link", "Reminder"]
    
    results = {}
    
    # Test each action
    for action in actions:
        
        # Copy the payment details
        payment_copy = payment_details.copy()
        
        # Add the action being tested
        payment_copy["recovery_action"] = action
        
        # Convert into DataFrame
        payment_df = pd.DataFrame([payment_copy])
        
        # Predict recovery probability
        probability = model.predict_proba(payment_df)[0][1]
        
        # Store result
        results[action] = probability
    
    # Find action with highest probability
    best_action = max(results, key=results.get)
    
    best_probability = results[best_action]
    
    return best_action, best_probability, results
```

```python
# AI Priority Function

def get_priority(payment_details, probability):

    amount = payment_details["amount"]

    # HIGH PRIORITY
    # High recovery probability + meaningful business value
    if probability >= 0.80 and amount >= 5000:
        return "HIGH"

    # High transaction value with reasonable recovery chance
    elif probability >= 0.60 and amount >= 20000:
        return "HIGH"

    # MEDIUM PRIORITY
    elif probability >= 0.50:
        return "MEDIUM"

    # LOW PRIORITY
    else:
        return "LOW"


def smart_recovery_agent(payment_details):

    # Available recovery actions
    actions = [
        "Retry",
        "UPI Link",
        "Reminder"
    ]

    # Safety rule:
    # If payment has already been retried 2 or more times,
    # do not recommend another retry
    if payment_details.get("retry_count", 0) >= 2:

        if "Retry" in actions:
            actions.remove("Retry")

    results = {}

    # Test every allowed recovery action
    for action in actions:

        # Create a copy so original payment data is not changed
        payment_copy = payment_details.copy()

        # Add current recovery action
        payment_copy["recovery_action"] = action

        # Convert to DataFrame for ML model
        payment_df = pd.DataFrame([
            payment_copy
        ])

        # Predict recovery probability
        probability = model.predict_proba(
            payment_df
        )[0][1]

        # Store probability for this action
        results[action] = float(probability)

    # Choose the action with highest probability
    best_action = max(
        results,
        key=results.get
    )

    best_probability = results[
        best_action
    ]

    # Calculate business priority
    priority = get_priority(
        payment_details,
        best_probability
    )

    return (
        best_action,
        best_probability,
        priority,
        results
    )
```

```python
import random
from datetime import datetime


def execute_recovery(payment_details):

    # Ask AI agent to choose the best action
    best_action, probability, priority, all_results = smart_recovery_agent(
        payment_details
    )

    # Convert probability to percentage
    confidence_percentage = round(
        probability * 100,
        2
    )

    # Expected recoverable revenue
    expected_recovery_amount = round(
        payment_details["amount"] * probability,
        2
    )

    # Generate AI reasoning
    reasoning = []

    reasoning.append(
        f"Predicted recovery probability is "
        f"{confidence_percentage}%."
    )

    reasoning.append(
        f"Recommended recovery action: {best_action}."
    )

    # Amount-based reasoning
    if payment_details["amount"] >= 10000:
        reasoning.append(
            "High transaction value increases business impact, "
            "so this payment should receive stronger attention."
        )

    # Historical customer behavior
    if payment_details.get(
        "previous_successful_payments", 0
    ) > payment_details.get(
        "previous_failed_payments", 0
    ):
        reasoning.append(
            "Customer has more successful payments than failed "
            "payments, indicating a higher likelihood of recovery."
        )

    # Retry reasoning
    if payment_details.get("retry_count", 0) >= 2:
        reasoning.append(
            "Multiple retry attempts were detected, so an "
            "alternative recovery method may be more suitable."
        )

    # Priority reasoning
    if priority == "HIGH":
        reasoning.append(
            "This payment is marked HIGH priority because of "
            "its recovery potential and business impact."
        )

    elif priority == "MEDIUM":
        reasoning.append(
            "This payment is marked MEDIUM priority and should "
            "be processed through the recommended recovery flow."
        )

    else:
        reasoning.append(
            "This payment is LOW priority because the predicted "
            "recovery potential is relatively limited."
        )

    # Create audit trail
    audit_trail = []

    current_time = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    # Payment failed event
    audit_trail.append({
        "time": current_time,
        "event": "Payment failed"
    })

    # AI analysis
    audit_trail.append({
        "time": current_time,
        "event": (
            f"AI analyzed payment and predicted "
            f"{confidence_percentage}% recovery probability"
        )
    })

    # Selected action
    audit_trail.append({
        "time": current_time,
        "event": (
            f"AI selected recovery action: "
            f"{best_action}"
        )
    })

    # Priority assigned
    audit_trail.append({
        "time": current_time,
        "event": (
            f"Payment assigned {priority} priority"
        )
    })

    # Simulate recovery based on ML probability
    recovered = random.random() < probability

    if recovered:

        recovered_amount = payment_details["amount"]

        status = "RECOVERED"

    else:

        recovered_amount = 0

        status = "NOT RECOVERED"

    # Final event
    audit_trail.append({
        "time": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),
        "event": (
            f"Recovery result: {status}"
        )
    })

    return {

        # AI decision
        "best_action": best_action,

        # ML prediction
        "recovery_probability": probability,

        "confidence_percentage": confidence_percentage,

        # Business priority
        "priority": priority,

        # Expected revenue
        "expected_recovery_amount":
            expected_recovery_amount,

        # Explainable AI reasoning
        "reasoning": reasoning,

        # Recovery result
        "status": status,

        "recovered_amount":
            recovered_amount,

        # Alternative predictions
        "all_predictions":
            all_results,

        # Complete event history
        "audit_trail":
            audit_trail
    }
```

```python
# process multiple payments 

# Process Multiple Payments

import pandas as pd
from datetime import datetime


def process_multiple_payments(payments):

    results = []

    for payment in payments:

        # -----------------------------------
        # Run complete AI recovery workflow
        # -----------------------------------

        result = execute_recovery(payment)


        # -----------------------------------
        # Current processing time
        # -----------------------------------

        processed_time = datetime.now()


        # -----------------------------------
        # Store payment + AI results
        # -----------------------------------

        results.append({

            # Basic payment information
            "amount": payment["amount"],

            "payment_method": payment["payment_method"],

            "failure_reason": payment["failure_reason"],


            # -----------------------------------
            # AI Decision
            # -----------------------------------

            "best_action": result["best_action"],


            # -----------------------------------
            # ML Prediction
            # -----------------------------------

            "recovery_probability": result[
                "recovery_probability"
            ],

            "confidence_percentage": result[
                "confidence_percentage"
            ],


            # -----------------------------------
            # Business Priority
            # -----------------------------------

            "priority": result["priority"],


            # -----------------------------------
            # Revenue Intelligence
            # -----------------------------------

            "expected_recovery_amount": result[
                "expected_recovery_amount"
            ],


            "recovered_amount": result[
                "recovered_amount"
            ],


            # -----------------------------------
            # Explainable AI
            # -----------------------------------

            "ai_reasoning": result.get(
                "reasoning",
                [
                    "AI analyzed payment behavior and "
                    "selected the best recovery strategy."
                ]
            ),


            # -----------------------------------
            # Alternative action predictions
            # -----------------------------------

            "all_predictions": result.get(
                "all_predictions",
                {}
            ),


            # -----------------------------------
            # Audit Trail
            # -----------------------------------

            "audit_trail": result.get(
                "audit_trail",
                []
            ),


            # -----------------------------------
            # Recovery Status
            # -----------------------------------

            "status": result["status"],


            # -----------------------------------
            # Timestamps
            # -----------------------------------

            "created_at": processed_time,


            "recovered_at": (
                processed_time
                if result["status"] == "RECOVERED"
                else None
            )

        })


    # -----------------------------------
    # Return all processed payments
    # -----------------------------------

    return pd.DataFrame(results)
```

```python

# BUSINESS-AWARE PRIORITY SCORE


# Normalize transaction amount between 0 and 1
max_amount = batch_results["amount"].max()

batch_results["normalized_amount"] = (
    batch_results["amount"] / max_amount
)

# Priority Score
# 70% importance -> recovery likelihood
# 30% importance -> financial impact

batch_results["business_priority_score"] = (
    0.70 * batch_results["recovery_probability"]
    +
    0.30 * batch_results["normalized_amount"]
)

batch_results[
    [
        "amount",
        "recovery_probability",
        "business_priority_score"
    ]
].head()
```

```python

# ASSIGN BUSINESS PRIORITY


def assign_business_priority(score):

    if score >= 0.70:
        return "HIGH"

    elif score >= 0.45:
        return "MEDIUM"

    else:
        return "LOW"


batch_results["business_priority"] = (
    batch_results["business_priority_score"]
    .apply(assign_business_priority)
)

batch_results[
    [
        "amount",
        "recovery_probability",
        "business_priority_score",
        "business_priority"
    ]
].head(10)
```

```python

# EXPLAINABLE AI DECISION REASONING


high_value_threshold = (
    batch_results["amount"].quantile(0.75)
)


def generate_ai_reasoning(row):

    reasons = []

    probability = row["recovery_probability"] * 100

    # Recovery likelihood
    if probability >= 80:

        reasons.append(
            f"High recovery likelihood ({probability:.1f}%)"
        )

    elif probability >= 50:

        reasons.append(
            f"Moderate recovery likelihood ({probability:.1f}%)"
        )

    else:

        reasons.append(
            f"Lower recovery likelihood ({probability:.1f}%)"
        )


    # Financial impact
    if row["amount"] >= high_value_threshold:

        reasons.append(
            f"High transaction value of ₹{row['amount']:,.0f} increases business impact"
        )


    # Recovery action reasoning
    if row["best_action"] == "Retry":

        reasons.append(
            "Retry is predicted to be the most effective recovery action"
        )

    elif row["best_action"] == "UPI Link":

        reasons.append(
            "UPI Link is predicted to provide a better alternative recovery path"
        )

    elif row["best_action"] == "Reminder":

        reasons.append(
            "Reminder is selected as the most suitable customer follow-up action"
        )


    # Final priority explanation
    if row["business_priority"] == "HIGH":

        reasons.append(
            "High expected business recovery opportunity"
        )

    elif row["business_priority"] == "MEDIUM":

        reasons.append(
            "Moderate recovery opportunity"
        )

    else:

        reasons.append(
            "Lower recovery opportunity compared with other failed payments"
        )


    return ". ".join(reasons) + "."


batch_results["ai_reasoning"] = (
    batch_results.apply(
        generate_ai_reasoning,
        axis=1
    )
)


batch_results[
    [
        "amount",
        "best_action",
        "recovery_probability",
        "expected_recovery_value",
        "business_priority",
        "ai_reasoning"
    ]
].head()
```

```python
# dashboard metrics

# RePay Dashboard Metrics

total_payments = len(batch_results)

# Total failed payment value
total_revenue_at_risk = (
    batch_results["amount"].sum()
)

# Number of recovered payments
recovered_payments = (
    batch_results["status"] == "RECOVERED"
).sum()

# Actual recovered revenue
total_revenue_recovered = (
    batch_results["recovered_amount"].sum()
)

# Recovery rate
recovery_rate = (
    recovered_payments / total_payments
) * 100 if total_payments > 0 else 0


# Expected recoverable revenue based on ML confidence
expected_recovery_revenue = (
    batch_results["amount"]
    * batch_results["recovery_probability"]
).sum()


# High priority payments
high_priority_count = (
    batch_results["priority"] == "HIGH"
).sum()


# Average AI confidence
average_recovery_probability = (
    batch_results["recovery_probability"].mean()
    * 100
)


print("RePay — AI Payment Recovery Dashboard")
print("=" * 50)

print(f"Total Failed Payments: {total_payments}")

print(
    f"Total Revenue at Risk: "
    f"₹{total_revenue_at_risk:,.2f}"
)

print(f"Recovered Payments: {recovered_payments}")

print(
    f"Revenue Recovered: "
    f"₹{total_revenue_recovered:,.2f}"
)

print(
    f"Expected Recoverable Revenue: "
    f"₹{expected_recovery_revenue:,.2f}"
)

print(
    f"Recovery Rate: "
    f"{recovery_rate:.2f}%"
)

print(
    f"High Priority Payments: "
    f"{high_priority_count}"
)

print(
    f"Average AI Recovery Confidence: "
    f"{average_recovery_probability:.2f}%"
)
```
